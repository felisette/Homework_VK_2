```python
import numpy as np
from collections import deque
```

```python
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None

def create_list(values):
    if not values:
        return None
    head = ListNode(values[0])
    tmp = head
    for v in values[1:]:
        tmp.next = ListNode(v)
        tmp = tmp.next
    return head

def print_list(head):
    if head is None:
        print("None")
        return
    tmp = head
    res = []
    while tmp is not None:
        res.append(str(tmp.val))
        tmp = tmp.next
    print(" ".join(res))
    return
```

### Задание 1 Является ли список циклическим

```python
def create_list_cycle(values, pos):
    if not values:
        return None
    head = ListNode(values[0])
    tmp = head
    cycle_node = None
    if pos == 0:
        cecle_node = head
    for i, v in enumerate(values[1:], start=1):
        tmp.next = ListNode(v)
        tmp = tmp.next
        if i == pos:
            cycle_node = tmp
    if cycle_node:
        tmp.next = cycle_node
    return head
    
def hasCycle(head):
    if head is None or head.next is None:
        return False
    slow = head
    fast = head.next
    while slow != fast:
        if fast is None or fast.next is None:
            return False
        slow = slow.next
        fast = fast.next.next
    return True
```

```python
val = [2, 4, 5, 3, 5, 7, 8, 9, 11, 0]
lst = create_list(val)
print(hasCycle(lst))
lst_with_cycle = create_list_cycle(val, pos = 5)
print(hasCycle(lst_with_cycle))
```

### Задание 2 Развернуть односвязный список

```python
def reverseLinkedList(head):
    prev = None
    tmp = head
    while tmp is not None:
        next_node = tmp.next
        tmp.next = prev
        prev = tmp
        tmp = next_node
    head = prev
    return head
```

```python
val = [2, 4, 5, 3, 5, 7, 8, 9, 11, 0]
lst = create_list(val)
print_list(lst)
reverse_list = reverseLinkedList(lst)
print_list(reverse_list)
```

### Задание 3 Найти середину списка

```python
def middleNode(head):
    slow = head
    fast = head
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
    return slow
```

```python
val = [2, 4, 5, 3, 5, 7, 8, 9, 11, 0]
lst = create_list(val)
m = middleNode(lst)
print(m.val)
```

```python
val = [2, 4, 5, 3, 5, 7, 8, 9, 11]
lst = create_list(val)
m = middleNode(lst)
print(m.val)
```

### Задание 4 Удалить элемент из односвязного списка

```python
def removeElements(head, val):
    dummy = ListNode(0)
    dummy.next = head
    prev = dummy
    tmp = head
    while tmp is not None:
        if tmp.val == val:
            prev.next = tmp.next
        else:
            prev = tmp
        tmp = tmp.next
    return dummy.next
```

```python
val = [2, 4, 5, 3, 5, 7, 8, 9, 11, 0]
lst = create_list(val)
print_list(lst)
lst2 = removeElements(lst, 11)
print_list(lst2)
```

```python
val = [2, 4, 5, 3, 5, 7, 8, 9, 11, 0]
lst = create_list(val)
print_list(lst)
lst2 = removeElements(lst, 19)
print_list(lst2)
```

### Задание 5 Удаление n-ого элемента с конца списка

```python
def removeNthFromEnd(head, n):
    dummy = ListNode(0)
    dummy.next = head
    slow = dummy
    fast = dummy
    for i in range(n+1):
        if fast.next is None:
            return head
        fast = fast.next
    while fast is not None:
        slow = slow.next
        fast = fast.next
    slow.next = slow.next.next
    return dummy.next
```

```python
val = [2, 4, 5, 3, 5, 7, 8, 9, 11, 0]
lst = create_list(val)
print_list(lst)
lst2 = removeNthFromEnd(lst, 20)
print_list(lst2)
```

### Задание 6 Является ли одна строка исходной для другой

```python
def isSubsequence(a, b):
    q = deque()
    for el in a:
        q.append(el)
    for el in b:
        if len(q)>0 and q[0] == el:
            q.popleft()
    return len(q) == 0
```

```python
a = "qwertyuiop"
b = "werty"
print(isSubsequence(a, b))
```

```python
a = "werty"
b = "qwertyuiop"
print(isSubsequence(a, b))
```

```python
a = ""
b = "qwerty"
print(isSubsequence(a, b))
print(isSubsequence(b, a))
```

### Задание 7 Является ли одна строка исходной для другой - метод двух указателей

```python
def isSubsequence(a, b):
    i = 0
    lenA = len(a)
    lenB = len(b)
    for el1 in range(lenA):
        for el2 in range(lenB):
            if a[el1] == b[el2]:
                i+=1
                break
    return i == lenA
```

```python
a = "qwertyuiop"
b = "werty"
print(isSubsequence(a, b))
```

```python
a = "werty"
b = "qwertyuiop"
print(isSubsequence(a, b))
```

```python
a = ""
b = "qwerty"
print(isSubsequence(a, b))
print(isSubsequence(b, a))
```

### Задание 8 Является ли слово палиндромом

```python
def isPalindrome(s):
    stack = []
    for char in s:
        stack.append(char)
    for char in s:
        if char!=stack.pop():
            return False
    return True
```

```python
s = "gsjdkde"
print(isPalindrome(s))
```

```python
s = ""
print(isPalindrome(s))
```

```python
s = "qwertyytrewq"
print(isPalindrome(s))
```

### Задание 9 Является ли слово палиндромом - метод двух указателей

```python
def isPalindrome2(s):
    left = 0
    right = len(s) - 1
    while left<right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True
```

```python
s = "gsjdkde"
print(isPalindrome2(s))
```

```python
s = ""
print(isPalindrome2(s))
```

```python
s = "qwertyytrewq"
print(isPalindrome2(s))
```

### Задание 10 Сортировка масива через дополнительный стек

```python
def sortArrayUsingStack(input_arr):
    stack = []
    while len(input_arr) > 0:
        current = input_arr.pop()
        while len(stack) > 0 and stack[-1] > current:
            input_arr.append(stack.pop())
        stack.append(current)
    return stack
```

```python
val = [2, 4, 5, 3, 5, 7, 8, 9, 11, 0]
print(sortArrayUsingStack(val))
```

### Задание 11 Слияние двух отсортированных списков

```python
def mergeList(lst1, lst2):
    dummy = ListNode(0)
    tail = dummy
    while lst1 is not None and lst2 is not None:
        if lst1.val <= lst2.val:
            tail.next = lst1
            lst1 = lst1.next
        else:
            tail.next = lst2
            lst2 = lst2.next
        tail = tail.next
    if lst1 is not None:
        tail.next = lst1
    else:
        tail.next = lst2
    return dummy.next
```

```python
val1 = [2, 4, 5, 7, 8, 9, 11]
val2 = [1, 3, 6, 10, 13, 15]
lst1 = create_list(val1)
lst2 = create_list(val2)
print_list(lst1)
print_list(lst2)
lst3 = mergeList(lst1, lst2)
print_list(lst3)
```
